# **Project Summary -**

The Google Play Store dataset, when first encountered, presented a rich collection of information regarding various mobile applications. It comprised a total of 10,841 rows and 13 distinct columns, each intended to capture a specific characteristic of the apps listed on the platform. The columns included: App, Category, Rating, Reviews, Size, Installs, Type, Price, Content Rating, Genres, Last Updated, Current Ver, and Android Ver.

Upon initial inspection of the data types, a predominant pattern emerged: twelve out of the thirteen columns were classified as object data types. This classification typically indicates that the data within these columns is stored as strings, or a mix of data types, rather than numerical or specific temporal formats. The sole exception was the Rating column, which was correctly identified as a float64 (a floating-point number). This observation immediately signaled a significant challenge for quantitative analysis, as several columns that inherently represent numerical values—such as Reviews, Size, Installs, and Price—were stored as objects. This meant that direct mathematical operations or statistical computations on these columns would not be possible without prior conversion.

Further examination revealed several notable issues and characteristics across individual columns:

Rating: While a float64, the describe() function showed a maximum value of 19.0, which is an outlier given that app ratings are typically on a scale of 1 to 5. More critically, this column contained a substantial number of missing values, totaling 1,474 entries.

Reviews: This column was stored as an object, despite representing numerical counts of user reviews. This implied the presence of non-numeric characters or inconsistencies within the entries.

Size: Also an object type, the Size column displayed a variety of formats. Entries included numerical values followed by 'M' for megabytes (e.g., '19M'), 'k' for kilobytes (e.g., '201k'), and a specific string 'Varies with device'. This heterogeneity made direct numerical interpretation impossible.

Installs: This critical metric, intended to represent the number of installations, was stored as an object. Unique values revealed the presence of special characters such as '+' (e.g., '10,000+') and commas (e.g., '5,000,000+'), along with numerical entries, all preventing direct numerical use.

Type: This column, indicating whether an app is 'Free' or 'Paid', was an object and contained one missing value.

Price: Stored as an object, this column's unique values showed monetary symbols (e.g., '$4.99') and also the string '0' for free apps. Notably, one entry was an erroneous 'Everyone', indicating a data entry error.

Content Rating: This categorical column, describing the appropriate audience for the app, was an object and had one missing value.

Current Ver: The app's current version, stored as an object, had 8 missing values.

Android Ver: The minimum Android version required, also an object, had 3 missing values.

## 3. ***Data Wrangling***

### Data Wrangling Code

````python
playdata.duplicated().value_counts()
playdata.shape
````

````python
#We have 483 rows of duplicates in the data set.
#we are dropping those duplicates to make it much better for analysis.
#Inplace is used to change the original data set.
playdata.drop_duplicates(inplace=True)
print(f" Previous rows was 10841 but now is : {playdata.shape[0]}")
````

````python
#There is one category which has given as 1.9. Compared to remaining data in that column this value is completely different
#May be entry error.
#We are going to remove that row
index_to_drop = playdata[playdata['Category'] == '1.9' ].index# We extracted the index of the unique value and
#mentioned it in the drop. so that it exactly removes the index we have given.
playdata.drop(index_to_drop, inplace = True)
print(f" After dropping the 'Category' that has '1,9' is : {playdata.shape}")
````

````python
#We should check wether Reviews column has only int or float.
#if not then we should convert them into int or float.
#here we use pandas function to_numeric() and check all are int or not.
#if not errors='coerce' will remove the non numeric and set it as NaN.
playdata['Reviews'] = pd.to_numeric(playdata['Reviews'], errors='coerce')
playdata.info()
````

````python
playdata["Size"].replace('Varies with device', np.nan,inplace = True)
# replaced 'varies with device' to NaN value and used inplace = True to modify original dataset
````

````python

#introduced custom function KtoM
def KtoM(s):
  if 'k' in str(s):
    #type casted s to str and checks str(s) carries any k value
    #if it carries then replace it with ''
    #then type cast back to float
    x = float(str(s).replace('k', ''))
    #to convert KB to MB we should divide the value with 1024
    #1MB = 1024KB and return
    return x / 1024
  elif 'M' in str(s):
    x = float(str(s) . replace('M', ''))
    #if no 'k' in then check for 'M'. if yes, remove the M and then type cast to float
    return x
  else:
    return float(s)
    #none of k or M present in row then just return float(s)


playdata['Size'] = playdata['Size'].apply(KtoM)
#applied custom function here to check all
playdata['Size'].sample(5)
````

````python
# Clean and Convert 'Installs' to Numeric: Clean the 'Installs' column by removing special characters like '+' and ',' and converting it to a numeric type.
playdata['Installs'] = playdata['Installs'].apply(lambda x: int(x.replace('+','').replace(',', '')) if '+' in str(x) else int(x))
#applied lambda and removing ',' and '+' and while doing to type cast it into int
playdata['Installs'] = pd.to_numeric(playdata['Installs'], errors = 'coerce')
#if other than numeric add NaN
playdata['Installs'].sample(5)
````

````python
#cleaning the price column by removing special character and type cast it into float
playdata['Price'] = playdata['Price'].apply(lambda x: float(str(x).replace('$','')))
````

````python
playdata['Content Rating'].value_counts()
````

````python
print("Missing values before imputation:")
#we need to get count of null values and try to fill those up with median or remove if ok.
print(playdata[['Rating', 'Type', 'Content Rating', 'Current Ver', 'Android Ver']].isnull().sum())
````

````python
#Creatng median() out of Rating and fill empty with Median()
Rating_median = playdata['Rating'].median()
#fill null with median
playdata['Rating'].fillna(Rating_median,inplace = True)
````

````python
mode_type = playdata['Type'].mode()[0]
#mode()[0] it returns a pandas Series containing the most frequent value in that column.
#we want to get first mode and then assign it to the null in type
playdata['Type'] = playdata['Type'].fillna(mode_type)
print(f"Missing values in 'Type' after imputation: {playdata['Type'].isnull().sum()}")
````

````python
#we do the same to the Current ver fill the nan with most frequent in the columnsz
mode_current_ver = playdata['Current Ver'].mode()[0]
playdata['Current Ver'] = playdata['Current Ver'].fillna(mode_current_ver)
print(f"Missing values in 'Current Ver' after imputation: {playdata['Current Ver'].isnull().sum()}")
````

````python
mode_android_ver = playdata['Android Ver'].mode()[0]
playdata['Android Ver'] = playdata['Android Ver'].fillna(mode_android_ver)
print(f"Missing values in 'Android Ver' after imputation: {playdata['Android Ver'].isnull().sum()}")
````

````python
print("Missing values after imputation:")
print(playdata[['Rating', 'Type', 'Content Rating', 'Current Ver', 'Android Ver']].isnull().sum())
````
